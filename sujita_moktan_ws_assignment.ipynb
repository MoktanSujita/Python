```python
print("Select any operation:\n"
         "1. ADD\n"
         "2. Subtract\n"
         "3. Multiply\n"
         "4. Divide\n")
#input request and acceptance with "input()"
ope = int(input("Select any operation from (1-4):"))    #typecasting the input using int function
n1 = int(input("Enter first number:"))
n2 = int(input("Enter second number:"))

#switch case
match ope:
  case 1:
   print(f"Sum:  {n1+n2}")     #using formatted string literal to concatenate the results
  case 2:
   print(f"Difference: {n1-n2}")
  case 3:
    print(f"Multiplicatio: {n1*n2}")
  case 4:
    print(f"Division: {n1/n2}")
  case _:                      #if no match is found, execute the following code   
     print("Please enter valid operation number")
   
```
